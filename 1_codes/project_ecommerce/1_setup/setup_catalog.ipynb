### Source data volume and checkpoints
Raw CSV files are uploaded to `/Volumes/ecommerce/source_data/raw/` (same folder layout as `0_data/ecomm-raw-data/`).
Auto Loader keeps track of already ingested order_items files in `/Volumes/ecommerce/source_data/checkpoints/`.

```python
%sql
CREATE SCHEMA IF NOT EXISTS ecommerce.source_data;
CREATE VOLUME IF NOT EXISTS ecommerce.source_data.raw;
CREATE VOLUME IF NOT EXISTS ecommerce.source_data.checkpoints;
```
